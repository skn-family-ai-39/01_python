# for 반복문
- 반복 횟수가 지정된 반복에 사용되는 구문
- python for문은 iterable 객체의 요소를 하나씩 꺼내 실행하는 구문이다.
(iterable: 반복할 수 있는, 순회할 수 있는)
  - iterable 객체: list, tuple, set, dict, range, str 등...

# 1. 문자열, 리스트, 튜플 순회
- 시퀀스 타입 객체 (index가 존재한다)
- index 단위로 순회

```python
text: str = 'hello world'
```

```python
for ch in text:  # text 문자열에서 한 글자씩 꺼내 ch에 대입
    print(ch)    # 모든 글자를 한 번씩 꺼내면 반복 종료

print('종료됨')
```

```python
# 리스트 순회
# 변수명: 타입힌트 = 값
lst: list[int] =  [10, 20, 30, 40]

for num in lst:
    print(num)

print('종료됨')
```

```python
# 튜플 순회
tpl = (100, 200, 300)
for num in tpl:
    print(num)
```
