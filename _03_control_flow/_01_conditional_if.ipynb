# 조건문 if
- 조건문은 특정 조건에 따라 실행할 코드를 선택하는 문법이다.
- `if` - 조건이 참인 경우에만 실행
- `if...else` - 둘 중 하나를 선택해서 실행
- `if...else...else` - 여러 조건 중 하나를 선택해서 실행

```python
# 함수 사용 예시
# - 선언, 정의, 호출의 3단계
# - 매개 변수 : 함수에 전달된 값을 저장하는 변수
# - 반환값 : 함수가 호출부로 돌려줄 값

def greeting(name): # name == 매개 변수       -함수 선언
    return name + "님 안녕하세요"  # 반환값    -함수 정의 

print(greeting('홍길동')) # - 함수 호출
```

```python
# give_coupon 함수 선언/정의
# 매개 변수 : int
# 반환값 : None

def give_coupon(score: int) -> None:   # 함수 선언

    # 점수가 90점 이상인 경우만 쿠폰 발급
    if score >= 90:
        print("쿠폰 발급")

user_input = int( input('점수 입력: ') )

give_coupon(user_input)  # 함수 호출
```

## 2. if ... else
- 조건이 참인 경우와 거짓인 경우를 나누어서 처리할 때 사용

```python
def pass_or_fail(score: int) -> None:
    if score >= 60:
        print(f"{score}점. 합격입니다")
    else:
        print(f"{score}점. 불합격입니다")

pass_or_fail(80)
pass_or_fail(50)
```

## 3. if ... elif ... else
- 여러 조건에 대한 동작을 제어할 때 사용

```python
# 점수를 입력하면 등급을 반환하는 함수
def get_grade(score: int) -> str:

    # 유효성 검사(validation)
    # - 제공 받은 값이 기능 수행에 올바른 값인지 확인

    # score가 0~100 사이 값이 아니면
    # == score가 0 미만 또는 100 초과인 경우
    if not(0 <= score <= 100):

        # 에러 강제 발생
        raise valueError("0~100 사이 값만 입력하세요")

    grade = None # 비어있는 변수

    if score >= 90:
        grade = 'A'
    elif score >= 80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60:
        grade = 'D'
    else:
        grade = 'F'
    
    return grade

print(get_grade(87))
print(get_grade(87))
print(get_grade(65))
print(get_grade(90))
print(get_grade(44))
print(get_grade(200))
```

## 4. 중첩 if 
- if문을 중첩해서 작성

```python
def get_grade2(score: int) -> str:

    if not(0 <= score <= 100):
        raise ValueError("0~100 사이 값만 입력")

    grade = None # 비어있는 변수

    if score >= 90:
        if score >= 95:
            grade = 'A+'
        else:
            grade = 'A'

    elif score >= 80: 
        if score >= 85:
            grade = 'B+'
        else:
            grade = 'B'

    elif score >= 70:
        grade = 'C'
    elif score >= 60: 
        grade = 'D'
    else:
        grade = 'F'
    
    return grade 


print( get_grade2(93) )
print( get_grade2(98) )
print( get_grade2(80) )
print( get_grade2(85) )
```

```python
# 개량 버전
def get_grade3(score: int) -> str:

    if not(0 <= score <= 100):
        raise ValueError("0~100 사이 값만 입력")

    grade = None # 비어있는 변수

    if score >= 90:
        grade = 'A'
    elif score >= 80: 
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60: 
        grade = 'D'
    else:
        grade = 'F'
    
    # score가 60점 이상이면서 1의 자리가 5 이상이면 + 추가

    # - score % 10  == 1의 자리만 남기기
    if score >= 60  and  score % 10 >= 5:
        grade = grade + '+'

    return grade 


print( get_grade2(93) )
print( get_grade2(98) )
print( get_grade2(80) )
print( get_grade2(85) )
print( get_grade2(55) )
```
